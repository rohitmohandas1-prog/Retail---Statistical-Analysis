```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Retail_Dataset_Cleaned.csv")

print("Dataset Loaded Successfully")
```

```python
df.head()
```

```python
df.info()
```

```python
revenue_variance = df["Revenue"].var()

print("Revenue Variance:", revenue_variance)
```

```python
import numpy as np

batsman_a =np.array([48, 50, 49, 51, 52, 50])
batsman_b = np.array([10, 90, 30, 70, 100, 0])

print("A -> Variance:", np.var(batsman_a), 'Std Dev:', np.std(batsman_a))
print("B -> Variance:", np.var(batsman_b), 'Std Dev:', np.std(batsman_b))
```

```python
revenue_std = df["Revenue"].std()

print("Revenue Standard Deviation:", revenue_std)
```

```python
comparison = pd.DataFrame({
    "Metric": ["Variance", "Standard Deviation"],
    "Value": [revenue_variance, revenue_std]
})

comparison
```

```python
units_std = df["Units_Sold"].std()

print("Units Sold Standard Deviation:", units_std)
```

```python
delivery_std = df["Delivery_Time"].std()

print("Delivery Time Standard Deviation:", delivery_std)
```
