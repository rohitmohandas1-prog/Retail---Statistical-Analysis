```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```
