## 1. GroupBy Basics

GroupBy splits your data into smaller groups based on a column, so you can study each group separately instead of the whole table at once.

**Real-life analogy:** Imagine 950 IPL matches all mixed together. Grouping by `Season` sorts every match into its own pile — a pile for 2019, a pile for 2020, and so on.

## 2. Aggregate Functions on Groups

Once grouped, we can run maths functions on each group: `sum()`, `mean()`, `count()`, `max()`, `min()`.

```python
movies.head(10)
```

```python
# mean() -> average rating per year
movies.groupby('year_of_release')['imdb_rating'].mean().head(8)
```

```python
# sum() -> total votes received per year, across all movies
movies.groupby('year_of_release')['imdb_votes'].sum().head(8)
```

```python
movies.head(1)
```

```python
# count() -> how many movies released each year
movies.groupby('year_of_release')['title_x'].count().head(8)
```

```python
# max() and min() -> highest and lowest rating per year
print(movies.groupby('year_of_release')['imdb_rating'].max().head(5))
print()
print(movies.groupby('year_of_release')['imdb_rating'].min().tail(5))
```

**Variation — same idea, but on the IPL dataset (matches won per team):**

```python
# size() here works like value_counts() -> total matches won by each team
ipl.groupby('WinningTeam').size().sort_values(ascending=False).head(8)
```

**Variation — average runtime of movies per year (runtime needs converting to numbers first):**

```python
movies['runtime'] = pd.to_numeric(movies['runtime'], errors='coerce')
movies.groupby('year_of_release')['runtime'].mean().head(8)
```

---
## 3. size() vs count() — What's the Real Difference?

`count()` skips missing (NaN) values. `size()` counts every row in the group, no matter what. Let's prove it using the IPL dataset's `City` column, which has real missing values.

```python
ipl['City'].isnull().sum()
```

```python
ipl
```

```python
# Compare count() and size() on the same column, same groups
comparison = pd.DataFrame({
    'count()': ipl.groupby('Season')['City'].count(),
    'size()': ipl.groupby('Season')['City'].size()
})
comparison[comparison['count()'] != comparison['size()']]
```

---
## 4. Single vs Multiple Column Aggregation — `agg()`

`agg()` lets you run several calculations at once, instead of writing separate lines for each one.

```python
# a) Multiple aggregations on ONE column -> pass a list
movies.groupby('year_of_release')['imdb_rating'].agg(['mean', 'sum', 'count','max','min']).head(5)
```

```python
# b) Different aggregations on DIFFERENT columns -> pass a dictionary
movies.groupby('year_of_release').agg({'imdb_rating': 'mean', 'imdb_votes': 'sum'}).head(6)
```

**Variation — agg() on the IPL dataset: number of matches AND number of unique winning teams, per season:**

```python
ipl.groupby('Season').agg({'ID': 'count', 'WinningTeam': 'nunique'}).head(8)
```

**Variation — agg() with a custom function using a lambda (range of ratings per year):**

```python
movies.groupby('year_of_release')['imdb_rating'].agg(lambda x: x.max() - x.min()).head(8)
```

## 6. pd.concat() — Stacking DataFrames Together

Use `concat()` when your data is split across multiple DataFrames and you just want to STACK them — as more rows, or as more columns.

```python
# a) Stacking ROWS together -> axis=0 (default)
batch1 = bat.iloc[0:3]      # pretend this came from File 1
batch2 = bat.iloc[3:6]      # pretend this came from File 2

pd.concat([batch1, batch2])
```

```python
# Understanding Ignore Index Parameter
import pandas as pd

batch1 = pd.DataFrame({
    "Name": ["Amit", "Rahul"]    
})

batch2 = pd.DataFrame({
    "Name": ["Priya", "Sneha"]
})

pd.concat([batch1, batch2],ignore_index=True)
```

```python
# Fixing the index so it doesn't repeat
pd.concat([batch1, batch2], ignore_index=True)
```

**Variation — using the `keys=` parameter to remember WHICH batch each row came from:**

```python
combined = pd.concat([batch1, batch2], keys=['File1', 'File2'])
combined
```

```python
# b) Joining COLUMNS together, side by side -> axis=1
top5 = bat.sort_values('batsman_run', ascending=False).head(5).reset_index(drop=True)
strike_rate = pd.DataFrame({'strike_rate': [139.8, 128.4, 143.9, 130.5, 136.7]})

pd.concat([top5, strike_rate], axis=1)
```

## 7. pd.merge() — Joining DataFrames Using a Common Column

`merge()` matches rows between two DataFrames using a COMMON column, like a SQL join. Let's build two small tables that DON'T perfectly match, so we can clearly see how each join type behaves.

```python
top_batsmen = bat.sort_values('batsman_run', ascending=False).head(5).reset_index(drop=True)
print(top_batsmen)

player_team = pd.DataFrame({
    'batter': ['V Kohli', 'S Dhawan', 'DA Warner', 'RG Sharma', 'MS Dhoni'],
    'Team': ['RCB', 'DC', 'SRH', 'MI', 'CSK']
})
print(player_team)
```

```python
# a) how='inner' (default) -> keep only rows that match in BOTH tables
pd.merge(top_batsmen, player_team, on='batter', how='inner')
```

```python
# b) how='left' -> keep every row from the LEFT table
pd.merge(top_batsmen, player_team, on='batter', how='left')
```

```python
# c) how='right' -> keep every row from the RIGHT table
pd.merge(top_batsmen, player_team, on='batter', how='right')
```

```python
# d) how='outer' -> keep every row from BOTH tables
pd.merge(top_batsmen, player_team, on='batter', how='outer')
```

**Variation — using `left_on` / `right_on` when the shared column has a different name:**

```python
player_team_v2 = player_team.rename(columns={'batter': 'PlayerName'})

pd.merge(top_batsmen, player_team_v2, left_on='batter', right_on='PlayerName', how='inner')
```

**Variation — using `suffixes` when both tables share a column name that ISN'T the join key:**

```python
season1 = pd.DataFrame({'batter': ['V Kohli', 'S Dhawan'], 'runs': [973, 689]})
season2 = pd.DataFrame({'batter': ['V Kohli', 'S Dhawan'], 'runs': [741, 502]})

pd.merge(season1, season2, on='batter', suffixes=('_2023', '_2024'))
```

## 🎯 Quick Recap

| Concept | What it does |
|---|---|
| `groupby('col')` | Splits data into groups based on a column's values |
| `sum()/mean()/count()/max()/min()` | Aggregate calculations applied to each group |
| `size()` vs `count()` | `size()` counts ALL rows; `count()` skips missing values |
| `agg([...])` / `agg({...})` | Multiple calculations at once — one column, or different ones per column |
| `groupby(['col1','col2'])` | Groups by more than one column |
| `pd.concat()` | Stacks DataFrames — more rows (axis=0) or more columns (axis=1) |
| `pd.merge()` | Joins two DataFrames using a shared column (inner/left/right/outer) |
