```python
df['Quantity_Available'] = pd.to_numeric(df['Quantity_Available'], errors='coerce')
```

```python
print(df['Quantity_Available'].dtypes)
```

```python
df['Units_Sold'] = pd.to_numeric(df['Units_Sold'], errors='coerce')
```

```python
print(df['Units_Sold'].dtypes)
```

```python
df['Shipping_Cost'] = pd.to_numeric(df['Shipping_Cost'], errors='coerce')
print(df['Shipping_Cost'].dtypes)
```

```python
print(df[['Quantity_Available', 'Units_Sold', 'Revenue', 'Shipping_Cost']].dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
missing_cols = df.isnull().sum()
print(
    missing_cols[missing_cols > 0]
)
```

```python
df[('City')].fillna(df['City'].mode()[0], inplace=True)
```

```python
df['Product_Name'].fillna(
    df['Product_Name'].mode()[0],
    inplace=True
)
```
